```python
def get_labels(img_names_lt):
    label_names_lt = []
    for img_name in img_names_lt:
        label_names_lt.append(os.path.splitext(img_name)[0] + '.txt')

    return label_names_lt

def copy_files(source_file_orig, source_path, destination_path):
    source_file = Path(source_path + source_file_orig)
    destination_dir = Path(destination_path)

    if destination_dir.exists():
        target_file = destination_dir / source_file.name
    else:
        target_file = Path('0')

    if not source_file.exists():
        print('Choose source file correctly')
    elif target_file.exists():
        print('Target file already exists')
    else:
        try:
            destination_dir.mkdir(parents=True, exist_ok=True)

            shutil.move(source_file, destination_dir)
        except PermissionError:
            print('Permission denied')
```

```python
train_labels = get_labels(train_imgs)
val_labels = get_labels(val_imgs)
test_labels = get_labels(test_imgs)
```
