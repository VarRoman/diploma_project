```python
import os
import glob
```

```python
def polygon_to_bbox(line):
    parts = line.strip().split()
    if len(parts) < 6 and parts[0] == '0':
        return line

    class_id = parts[0]
    coords = [float(x) for x in parts[1:]]

    xs = coords[0::2]
    ys = coords[1::2]

    x_min, x_max = min(xs), max(xs)
    y_min, y_max = min(ys), max(ys)

    width = x_max - x_min
    height = y_max - y_min
    center_x = x_min + width / 2
    center_y = y_min + height / 2

    return f"{class_id} {center_x:.6f} {center_y:.6f} {width:.6f} {height:.6f}"

def convert_dataset(labels_dir):
    txt_files = glob.glob(os.path.join(labels_dir, '*.txt'))

    if not txt_files:
        print(f"Помилка: Не знайдено .txt файлів у папці {labels_dir}")
        return

    count = 0
    for txt_file in txt_files:
        with open(txt_file, 'r') as f:
            lines = f.readlines()

        new_lines = []
        for line in lines:
            if line.strip():
                new_lines.append(polygon_to_bbox(line))

        with open(txt_file, 'w') as f:
            f.write('\n'.join(new_lines) + '\n')

        count += 1

    print(f"Успіх! Конвертовано {count} файлів розмітки у папці {labels_dir}.")
```

```python
train_labels_path = '/home/var-roman/Desktop/my_projects/diploma_project/data/pasha_v1/train/labels'
# val_labels_path = '/workspace/diploma_project/data/valid/labels'

if __name__ == "__main__":
    print("Починаємо конвертацію Train...")
    convert_dataset(train_labels_path)
```
